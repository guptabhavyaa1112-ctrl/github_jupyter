<h1> Data preprocessing - handling missing data,feature engineering etc</h1>

```python
#handle missing data 
df["Income"]=df["Income"].fillna(df["Income"].median())
```

<h1>Feature engineering</h1>

```python
#feature engineering
df.columns
```

```python
#age
df["age"]=2026-df["Year_Birth"]
```

```python
df.head()
```

```python
#customer tenure  
df["Dt_Customer"]=pd.to_datetime(df["Dt_Customer"],dayfirst=True)
reference_date=df["Dt_Customer"].max()
df["customer_tenure_days"]=(reference_date-df["Dt_Customer"]).dt.days
```

```python
df.head()
```

```python
df.columns
```

```python
#spending 
df["total_spending"]=df["MntWines"]+df["MntMeatProducts"]+df["MntFishProducts"]+df["MntSweetProducts"]+df["MntFruits"]+df["MntGoldProds"]
```

```python
df.head()
```

```python
#children 
df["total_children"]=df["Kidhome"]+df["Teenhome"]
```

```python
df.head()
```

```python
#education
df["Education"].value_counts()
```

```python
#undergrad,postgrad,grad
df["Education"]=df["Education"].replace({
    "Basic":"undergrad","2n Cycle":"undergrad","Graduation":"grad","Master":"postgrad","PhD":"postgrad"
})
```

```python
df["Education"].value_counts()
```

```python
#marital status 
df["Marital_Status"].value_counts()
```

```python
df["Living_with"]=df["Marital_Status"].replace({
    "Married":"partner","Together":"partner","Single":"alone","Divorced":"alone","Widow":"alone","Alone":"alone","Absurd":"alone","YOLO":"alone"
})
```

```python
df["Living_with"].value_counts()
```

<h1>Encoding</h1>

```python
df_cleaned.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe=OneHotEncoder()
cat_cols=["Education","Living_with"]
enc_cols=ohe.fit_transform(df_cleaned[cat_cols])
```

```python
enc_df=pd.DataFrame(enc_cols.toarray(),columns=ohe.get_feature_names_out(cat_cols),index=df_cleaned.index)
```

```python
enc_df.head()
```

```python
df_encoded=pd.concat([df_cleaned.drop(columns=cat_cols),enc_df],axis=1)
```

```python
df_encoded.shape
```

```python
df_encoded.head()
```

<h1>Scaling</h1>

```python
from sklearn.preprocessing import StandardScaler 
```

```python
X=df_encoded
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

<h1>Visualize</h1>

```python
#pca 
from sklearn.decomposition import  PCA
pca=PCA(n_components=3)
X_pca=pca.fit_transform(X_scaled)
```

```python
#plot
fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection="3d")
ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2])
ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
ax.set_title("3d projection")
```

<h1>Elbow method </h1>

<h1>analyze k value</h1>

```python
from sklearn.cluster import KMeans
from kneed import KneeLocator
```

```python
wcss=[]
for k in range(1,11):
    kmeans=KMeans(n_clusters=k,random_state=42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
```

```python
knee=KneeLocator(range(1,11),wcss,curve="convex",direction="decreasing")
optimal_k=knee.elbow
```

```python
print("optimal k(no .of clusters) :",optimal_k) 
```

```python
#plot
plt.plot(range(1,11),wcss,marker="o")
plt.xlabel("K")
plt.ylabel("wcss")
```

```python
#silhouette score
from sklearn.metrics import silhouette_score
scores=[]
for k in range(2,11):
    kmeans=KMeans(n_clusters=k,random_state=42)
    labels=kmeans.fit_predict(X_pca)
    score=silhouette_score(X_pca,labels)
    scores.append(score)
#plot
plt.plot(range(2,11),scores,marker="o")
plt.xlabel("K")
plt.ylabel("silhouette")
```

```python
#combined plot 
k_range=range(2,11)
fig,ax1=plt.subplots(figsize=(8,6))
ax1.plot(k_range,wcss[:len(k_range)],marker="o",color="blue")
ax1.set_xlabel("K")
ax1.set_ylabel("wcss")
ax2=ax1.twinx()
ax2.plot(k_range,scores[:len(k_range)],marker="x",color="red",linestyle="--")
ax2.set_ylabel("SS")
```

<h1>Clustering</h1>

```python
#kmeans
kmeans=KMeans(n_clusters=4,random_state=42)
labels_kmeans=kmeans.fit_predict(X_pca)
```

```python
fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection="3d")
ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2],c=labels_kmeans)
```

<h1>Best option in terms of clustering : agglomerative</h1>

```python
#agglomerative clustering 
from sklearn.cluster import AgglomerativeClustering 
```

```python
agg_clf=AgglomerativeClustering(n_clusters=4,linkage="ward")
labels_agg=agg_clf.fit_predict(X_pca)
```

```python
fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection="3d")
ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2],c=labels_agg)
```

<h1>Characterization of clusters </h1>

```python
df_cleaned["labels"]=labels_agg
```
